```python
import string
import time

myString = "long string that we are going to use to test the line wrap functionality of the print function in Python. It should wrap around to the next line when it reaches the end of the screen."
alphabet = string.ascii_lowercase
finished = ""

for targetLetter in myString.lower():
    if targetLetter == " ":
        finished += " "
        continue

    for alphabetLetter in alphabet:
        print("\r" + finished + alphabetLetter, end="", flush=True)
        time.sleep(0.04)

        if alphabetLetter == targetLetter:
            finished += targetLetter
            break

print()  # Move to the next line when finished
```
