## Exercise 3: Loop through a list

1. Create `scores = [75, 88, 92, 60, 81]`.
2. Use a `for` loop to print each score.
3. In another loop, print only scores greater than or equal to 80.
4. Calculate and print the total score.

```python
# Write your solution here
print("Loop through list")
scores = [75, 88, 92, 60, 81]
for score in scores:
    print(score)

print ("Scores greater than or equal to 80:")    
for score in scores:
    if score >= 80:
        print(score)
        
total_score = sum(scores)
print ("Total score:", total_score)
```

## Exercise 4: Comprehension and sort

1. From `scores = [75, 88, 92, 60, 81]`, create `high_scores` with only scores >= 80 using list comprehension.
2. Sort `high_scores` in ascending order.
3. Sort it again in descending order.
4. Print all results.

```python
# Write your solution here
scores = [75, 88, 92, 60, 81]

high_scores = [score for score in scores if score >= 80]

print(high_scores)

high_scores.sort()
print("ASC Order", high_scores)

high_scores.sort(reverse=True)
print("DESC Order", high_scores)
```
