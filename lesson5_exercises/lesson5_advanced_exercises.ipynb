## Exercise 1: Tuple unpacking patterns

- Create a tuple with at least 7 values.
- Unpack into `first`, `second`, `*middle`, `last`.
- Print each variable and lengths where relevant.

```python
# Write your solution here
values = (10, 20, 30, 40, 50, 60, 70)
first, second, *middle, last = values
print("First:", first)
print("Second:", second)
print("Middle:", middle)
print("Number of middle values:", len(middle))
print("Last:", last)
print("Total values:", len(values))
```

## Exercise 2: Tuple transformation

- Start from a tuple of integers.
- Create a new tuple with each value multiplied by 10.
- Print original and transformed tuples.

```python
# Write your solution here
numbers = (1, 2, 3, 4, 5)
multiplied_numbers = tuple(number * 10 for number in numbers)
print("Original:", numbers)
print("Transformed:", multiplied_numbers)
```

## Exercise 3: Set-based analytics

- Given two sets of course names, print union, intersection, and both differences.
- Print symmetric difference and explain what it means in a comment.

```python
# Write your solution here
student_one_courses = {"Python", "Databases", "Web Development"}
student_two_courses = {"Python", "Networking", "Web Development"}
print("Union:", student_one_courses | student_two_courses)
print("Intersection:", student_one_courses & student_two_courses)
print("Only student one:", student_one_courses - student_two_courses)
print("Only student two:", student_two_courses - student_one_courses)
# Symmetric difference contains courses taken by exactly one student.
print("Symmetric difference:", student_one_courses ^ student_two_courses)
```

## Exercise 4: Set cleanup task

- Create a set from a list containing repeated and mixed-case words.
- Normalize words to lowercase before inserting into set.
- Print final normalized unique set sorted alphabetically.

```python
# Write your solution here
words = ["Python", "JAVA", "python", "Java", "HTML", "html", "CSS"]
normalized_words = {word.lower() for word in words}
print(sorted(normalized_words))
```

## Exercise 5: Set from a list comprehension

- Use a list comprehension to build the squares of numbers 1..20 where the square is even.
- Convert the list to a set to remove duplicates (if any).
- Print the resulting set and its length.
- Convert it to a sorted list and print.

```python
# Write your solution here
even_squares = [number ** 2 for number in range(1, 21) if number ** 2 % 2 == 0]
unique_even_squares = set(even_squares)
print("Set:", unique_even_squares)
print("Length:", len(unique_even_squares))
print("Sorted list:", sorted(unique_even_squares))
```
