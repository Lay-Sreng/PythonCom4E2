## Exercise 1: Tuple basics

1. Create `colors = ("red", "green", "blue", "yellow")`.
2. Print the first item and last item.
3. Print the length of the tuple.
4. Check if `"green"` is in the tuple.

````python
colors = ("red", "green", "blue", "yellow")

print("First item:", colors[0])
print("Last item:", colors[-1])
print("Length:", len(colors))
print("Is green in the tuple?", "green" in colors)
````

## Exercise 2: Update a tuple (convert to list)

1. Start with `numbers = (10, 20, 30)`.
2. Convert it to a list.
3. Add `40`.
4. Convert it back to a tuple.
5. Print the final tuple.

````python
numbers = (10, 20, 30)
number_list = list(numbers)
number_list.append(40)
numbers = tuple(number_list)

print(numbers)
````

## Exercise 3: Loop and join tuples

1. Create `t1 = ("A", "B")` and `t2 = ("C", "D")`.
2. Loop through `t1` and print each item.
3. Join `t1` and `t2` into `t3`.
4. Print `t3`.

````python
t1 = ("A", "B")
t2 = ("C", "D")

for item in t1:
    print(item)

t3 = t1 + t2
print(t3)
````

## Exercise 4: Set basics

1. Create `items = {"pen", "book", "bag"}`.
2. Add `"pencil"`.
3. Update the set with `"eraser"` and `"ruler"`.
4. Remove one item using `remove()`.
5. Print the set after each change.

````python
# Write your solution here
items = {"pen", "book", "bag"}
print("Original set:", items)

items.add("pencil")
print("After add:", items)

items.update({"eraser", "ruler"})
print("After update:", items)

items.remove("pen")
print("After remove:", items)
````

## Exercise 5: Set operations

Given:

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
```

1. Find and print the union.
2. Find and print the intersection.
3. Find and print the difference `a - b`.
4. Find and print the symmetric difference.

````python
# Write your solution here
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference (a - b):", a - b)
print("Symmetric difference:", a ^ b)
````
